## Feature Engineering

```python
df['Total_Stops'].unique(), '-------------------' , df['Total_Stops'].mode()
```

```python
df['Total_Stops'] = df['Total_Stops'].map({'non-stop':0, '1 stop': 1, '2 stops': 2, '3 stops': 3, '4 stops': 4, np.nan : 1})
df['Total_Stops']
```

```python
df.head()
```

```python
# Date of journey
df['date'] = df['Date_of_Journey'].str.split('/').str[0].astype(int)
df['month'] = df['Date_of_Journey'].str.split('/').str[1].astype(int)
df['year'] = df['Date_of_Journey'].str.split('/').str[2].astype(int)

df.head()
```

```python
df.drop('Date_of_Journey', axis=1, inplace=True)
df.info()
```

```python
# duration
df['duration_hours'] = df['Duration'].str.split(' ').str[0].str.split('h').str[0]
df['duration_hours'].value_counts()
```

```python
df.loc[df['duration_hours'] == '5m', 'duration_hours'] = 0
df['duration_hours'] = df['duration_hours'].astype(int)
```

```python
df['duration_mins'] = df['Duration'].str.split(' ').str[1].str.split('m').str[0]
df['duration_mins'] = df['duration_mins'].fillna(0)

df['duration_mins'] = df['duration_mins'].astype(int)
```

```python
df.info()
```

```python
df.head()
```

```python
df['total_duration'] = df['duration_hours'] + df['duration_mins']/60
df.head(3)
```

```python
df.drop(['duration_hours', 'duration_mins', 'Duration'], axis=1, inplace=True)
```

```python
df
```

```python
# Departure Time
df['Dep_hour'] = df['Dep_Time'].str.split(':').str[0]
df['Dep_min'] = df['Dep_Time'].str.split(':').str[1]
```

```python
df['Dep_hour'] = df.Dep_hour.astype(int)
df['Dep_min'] = df.Dep_min.astype(int)
```

```python
df.info()
```

```python
df.drop(['Dep_Time'], inplace=True, axis=1)
```

```python
# arrival time
df.head()
```

```python
df['arrival_hour'] = df['Arrival_Time'].str.split(' ').str[0].str.split(':').str[0].astype(int)
df['arrival_min'] = df['Arrival_Time'].str.split(' ').str[0].str.split(':').str[1].astype(int)
```

```python
df.drop(['Arrival_Time'], inplace=True, axis=1)
df.head()
```

```python
# categorical features
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder()

df2 = pd.DataFrame(encoder.fit_transform(df[['Airline', 'Source', 'Destination']]).toarray(), 
                   columns=encoder.get_feature_names_out())

df2
```

```python
df_ready = pd.concat([df, df2], axis=1)
df_ready.head()
```

```python
df_ready.drop(['Airline', 'Source','Destination'], axis=1, inplace=True)
```

```python
df_ready.drop(['Additional_Info'], axis=1, inplace=True)
```
