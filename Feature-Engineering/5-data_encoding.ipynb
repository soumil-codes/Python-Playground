# Data Encoding

### one hot encoding

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.DataFrame({
    'color' : ['red', 'green', 'blue', 'green', 'green', 'red']
})
df
```

```python
# create and instance of OHE
encoder = OneHotEncoder()
```

```python
# perform fit and transform
encoded = encoder.fit_transform(df[['color']]).toarray()
```

```python
type(df[['color']])
```

```python
type(df['color'])
```

```python
encoded_df = pd.DataFrame(encoded, columns=encoder.get_feature_names_out())
encoded_df
```

```python
pd.concat([df, encoded_df], axis=1)
```

```python
import seaborn as sns
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
print(tips['sex'].value_counts(), "\n")
print(tips['smoker'].value_counts(), "\n")
print(tips['day'].value_counts(), "\n")
print(tips['time'].value_counts(), "\n")
```

```python
encoder_gender = OneHotEncoder()
encoded_gender = encoder_gender.fit_transform(tips[['sex']]).toarray()
```

```python
encoder_smoker = OneHotEncoder()
encoded_smoker = encoder_smoker.fit_transform(tips[['smoker']]).toarray()
```

```python
encoder_day = OneHotEncoder()
encoded_day = encoder_day.fit_transform(tips[['day']]).toarray()
```

```python
encoder_time = OneHotEncoder()
encoded_time = encoder_time.fit_transform(tips[['time']]).toarray()
```

```python
# tips_encoded = pd.DataFrame({
#     encoded_gender,
#     encoded_smoker,
#     encoded_day,
#     encoded_time
# }, columns=[
#     encoder_gender.get_feature_names_out(),
#     encoder_smoker.get_feature_names_out(),
#     encoder_day.get_feature_names_out(),
#     encoder_time.get_feature_names_out(),
# ]) wrong method

import numpy as np

tips_encoded = pd.DataFrame(
    np.hstack([
        encoded_gender,
        encoded_smoker,
        encoded_day,
        encoded_time
    ]),
    columns=np.concatenate([
        encoder_gender.get_feature_names_out(),
        encoder_smoker.get_feature_names_out(),
        encoder_day.get_feature_names_out(),
        encoder_time.get_feature_names_out()
    ])
)

tips_encoded.head()

```

```python
# best approach - encode all features in one go

encoder = OneHotEncoder(sparse_output=False)
categorical_cols = ['sex', 'smoker', 'day', 'time']
```

```python
encoded = encoder.fit_transform(tips[categorical_cols])
```

```python
tips_encoded_better = pd.DataFrame(encoded,
                                columns=encoder.get_feature_names_out(categorical_cols))

tips_encoded_better.head(10)
```

```python
whole_tips = pd.concat([tips[['total_bill', 'tip']],tips_encoded_better, tips['size']], axis=1)
whole_tips
```

### Ordinal encoding

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
df2 = pd.DataFrame({
    'size' : ['small', 'big', 'med', 'small', 'med', 'small']
})
df2
```

```python
## create and instance of ordinal encoder and then fit transform
encoder = OrdinalEncoder(categories=[['small', 'med', 'big']])
```

```python
encoder.fit_transform(df2[['size']])
```

```python
encoder.transform([['small']])
```
