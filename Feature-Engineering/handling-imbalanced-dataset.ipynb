# handling imabalanced data

### Generating data and setting up

```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
n = 1000
class_ratio = 0.9

n_class_0 = int(n*class_ratio)
n_class_1 = int(n-n_class_0)

n_class_0,n_class_1
```

```python
# Creating dataset

class0 = pd.DataFrame({
    'feature_1' : np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature_2' : np.random.normal(loc=0, scale=1, size=n_class_0),
    'target' : np.zeros(dtype=int, shape=(n_class_0))
})

print(class0.head())

class1 = pd.DataFrame({
    'feature_1' : np.random.normal(loc=0, scale=1, size=n_class_1),
    'feature_2' : np.random.normal(loc=0, scale=1, size=n_class_1),
    'target' : np.ones(dtype=int, shape=(n_class_1))
})

class1.head()
```

```python
class0.target.size, class1.target.size
```

```python
df = pd.concat([class0, class1]).reset_index(drop=True)
display(df.head()), display(df.tail())
```

```python
df_minority = df[df['target'] == 1]
df_minority
```

```python
df_majority = df[df['target'] == 0]
df_majority
```

```python
len(df_majority)
```

### Upscaling - increasing the minority

```python
from sklearn.utils import resample
```

```python
minorty_upscaled = resample(df_minority, replace=True, n_samples=len(df_majority))
minorty_upscaled.target.size
```

```python
df_upscaled = pd.concat([df_majority, minorty_upscaled]).reset_index(drop=True)
df_upscaled
```

### Downsampling - reducing the majority

```python
majority_downsampled = resample(df_majority,
                                replace=False,
                                n_samples=len(df_minority))
majority_downsampled.target.size
```

```python
df_downsampled = pd.concat([majority_downsampled, df_minority]).reset_index(drop=True)
df_downsampled
```
