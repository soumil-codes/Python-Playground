## About Dataset
Context
This is a small subset of dataset of Book reviews from Amazon Kindle Store category.

Content
5-core dataset of product reviews from Amazon Kindle Store category from May 1996 - July 2014. Contains total of 982619 entries. Each reviewer has at least 5 reviews and each product has at least 5 reviews in this dataset.
Columns

- asin - ID of the product, like B000FA64PK
- helpful - helpfulness rating of the review - example: 2/3.
- overall - rating of the product.
- reviewText - text of the review (heading).
- reviewTime - time of the review (raw).
- reviewerID - ID of the reviewer, like A3SPTOKDG7WBLN
- reviewerName - name of the reviewer.
- summary - summary of the review (description).
- unixReviewTime - unix timestamp.

```python
# Load the dataset
import pandas as pd
data=pd.read_csv('Kindle Reviews/all_kindle_review.csv')
data.head()
```

```python
df=data[['reviewText','rating']]
df.head()
```

```python
## Missing Values
df.isnull().sum()
```

```python
df['rating'].value_counts()
```

```python
## postive review is 1 and negative review is 0
df['rating']=df['rating'].apply(lambda x:0 if x<4 else 1)
```

```python
df['rating'].value_counts()
```

```python
## 1. Lower All the cases
df['reviewText']=df['reviewText'].str.lower()
```

```python
df.head()
```

```python
import re
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
```

```python
## Removing special characters
df['reviewText']=df['reviewText'].apply(lambda x:re.sub('[^a-z A-z 0-9-]+', '',x))
## Remove the stopswords
df['reviewText']=df['reviewText'].apply(lambda x:" ".join([y for y in x.split() if y not in stop_words]))
## Remove url 
df['reviewText']=df['reviewText'].apply(lambda x: re.sub(r'(http|https|ftp|ssh)://([\w_-]+(?:(?:\.[\w_-]+)+))([\w.,@?^=%&:/~+#-]*[\w@?^=%&/~+#-])?', '' , str(x)))
## Remove any additional spaces
df['reviewText']=df['reviewText'].apply(lambda x: " ".join(x.split()))
```

```python
## Lemmatizer
from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
def lemmatize_words(text):
    return " ".join([lemmatizer.lemmatize(word) for word in text.split()])
```

```python
df['reviewText']=df['reviewText'].apply(lambda x:lemmatize_words(x))
```

```python
## Train Test Split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(df['reviewText'],df['rating'],test_size=0.20)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
bow=CountVectorizer()
X_train_bow=bow.fit_transform(X_train).toarray()
X_test_bow=bow.transform(X_test).toarray()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer()
X_train_tfidf=tfidf.fit_transform(X_train).toarray()
X_test_tfidf=tfidf.transform(X_test).toarray()
```

```python
from sklearn.naive_bayes import GaussianNB
nb_model_bow=GaussianNB().fit(X_train_bow,y_train)
nb_model_tfidf=GaussianNB().fit(X_train_tfidf,y_train)
```

```python
y_pred_bow=nb_model_bow.predict(X_test_bow)
y_pred_tfidf=nb_model_tfidf.predict(X_test_tfidf)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Confusion Matrix for Bag of Words")
print(confusion_matrix(y_test,y_pred_bow))
print("Classification Report for Bag of Words")
print(classification_report(y_test,y_pred_bow))

print("Confusion Matrix for TF-IDF")
print(confusion_matrix(y_test,y_pred_tfidf))
print("Classification Report for TF-IDF")
print(classification_report(y_test,y_pred_tfidf))
```

```python
print("Accuracy for Bag of Words:",accuracy_score(y_test,y_pred_bow))
print("Accuracy for TF-IDF:",accuracy_score(y_test,y_pred_tfidf))
```

```python
import gensim 
from gensim.models import Word2Vec, keyedvectors
```

```python
import gensim.downloader as api
w2v=api.load('word2vec-google-news-300')
```

```python
sentences = df['reviewText'].apply(lambda x: x.split()).tolist()
type(sentences), type(sentences[0]), type(sentences[0][0])
```

```python
import numpy as np
```

```python
def avgWord2VEC(sentence):
    M = []
    for word in sentence:
        try:
            M.append(w2v[word])
        except:
            pass
    M = np.array(M)
    v = M.mean(axis=0)
    if type(v) != np.ndarray:
        return np.zeros(300)
    return v
```

```python
X = np.array([avgWord2VEC(x) for x in sentences])
```

```python
X.shape
```

```python
X_train_word2vec, X_test_word2vec, y_train_word2vec, y_test_word2vec = train_test_split(X, df['rating'], test_size=0.20)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter=1000)
model.fit(X_train_word2vec, y_train_word2vec)
y_pred_word2vec = model.predict(X_test_word2vec)
```

```python
print(accuracy_score(y_test_word2vec, y_pred_word2vec))
print(classification_report(y_test_word2vec, y_pred_word2vec))
```
