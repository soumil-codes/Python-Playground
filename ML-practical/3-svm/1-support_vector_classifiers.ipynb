# SVC - SUPPORT VECTOR CLASSIFIER

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import make_classification
```

```python
X,y = make_classification(n_samples=1000, n_clusters_per_class=1, n_features=2, n_classes=2, n_redundant=0)
```

```python
X
```

```python
sns.scatterplot(x=pd.DataFrame(X)[0], y=pd.DataFrame(X)[1], hue=y)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
from sklearn.svm import SVC
```

```python
linear = SVC(kernel='linear')
```

```python
linear.fit(X_train, y_train)
```

```python
y_pred = linear.predict(X_test)
y_pred
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```

since the data was completely linearly separable we got a accuracy of 100%

```python
X,y = make_classification(n_samples=1000, n_clusters_per_class=2, n_features=2, n_classes=2, n_redundant=0)
sns.scatterplot(x=pd.DataFrame(X)[0], y=pd.DataFrame(X)[1], hue=y)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
linear.fit(X_train, y_train)
y_pred = linear.predict(X_test)
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```

```python
rbf = SVC(kernel='rbf')
rbf.fit(X_train, y_train)
y_pred2 = rbf.predict(X_test)
print(classification_report(y_test, y_pred2))
print(confusion_matrix(y_test, y_pred2))
```

```python
poly = SVC(kernel='poly')
poly.fit(X_train, y_train)
y_pred3 = poly.predict(X_test)
print(classification_report(y_test, y_pred3))
print(confusion_matrix(y_test, y_pred3))
```

```python
sigmoid = SVC(kernel='sigmoid')
sigmoid.fit(X_train, y_train)
pred3 = sigmoid.predict(X_test)
print(classification_report(y_test, pred3))
print(confusion_matrix(y_test, pred3))
```

```python
from sklearn.model_selection import GridSearchCV
gid_params = {
    'kernel' : ['rbf', 'poly', 'linear'],
    'C' : [0.1, 1, 10, 100, 1000],
    'gamma' : [0.1, 0.01, 0.001, 0.0001]
}
```

```python
grid = GridSearchCV(SVC(), param_grid=gid_params, cv=5, verbose=3, refit=True)
```

```python
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
pred4 = grid.predict(X_test)
print(classification_report(y_test, pred4))
print(confusion_matrix(y_test, pred4))
```
